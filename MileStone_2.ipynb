```python
!pip install PyPDF2 pdfplumber python-docx

import PyPDF2, pdfplumber, docx

def extract_text_from_txt(path):
    with open(path, "r", encoding="utf-8") as f:
        return f.read()

def extract_text_from_docx(path):
    doc = docx.Document(path)
    return "\n".join([para.text for para in doc.paragraphs])

def extract_text_from_pdf_pypdf2(path):
    text = ""
    with open(path, "rb") as f:
        reader = PyPDF2.PdfReader(f)
        for page in reader.pages:
            text += page.extract_text() or ""
    return text

def extract_text_from_pdf_pdfplumber(path):
    text = ""
    with pdfplumber.open(path) as pdf:
        for page in pdf.pages:
            text += page.extract_text() or ""
    return text
```

```python
# Install required packages (run this first)
!sudo apt install tesseract-ocr
!pip install pytesseract pillow pdfplumber

import pytesseract
from PIL import Image
import pdfplumber
import os

def extract_with_ocr(pdf_path):
    """Extract text from scanned PDF using OCR"""
    text = ""
    try:
        with pdfplumber.open(pdf_path) as pdf:
            print(f"Processing {len(pdf.pages)} pages with OCR...")

            for i, page in enumerate(pdf.pages, 1):
                print(f"  - Analyzing page {i}...")

                # Convert page to image and perform OCR
                img = page.to_image(resolution=300).original
                page_text = pytesseract.image_to_string(img)
                text += f"\n--- Page {i} ---\n{page_text}\n"

        return text
    except Exception as e:
        return f"Error during OCR processing: {str(e)}"

def extract_text_from_pdf_pdfplumber(pdf_path):
    """Extract text from searchable PDF (non-scanned)"""
    text = ""
    try:
        with pdfplumber.open(pdf_path) as pdf:
            print(f"Extracting text from {len(pdf.pages)} pages...")

            for i, page in enumerate(pdf.pages, 1):
                page_text = page.extract_text()
                if page_text:
                    text += f"\n--- Page {i} ---\n{page_text}\n"
                else:
                    text += f"\n--- Page {i} (No text found, might be scanned) ---\n"

        return text
    except Exception as e:
        return f"Error during PDF extraction: {str(e)}"

def smart_pdf_extraction(pdf_path):
    """Smart extraction that tries regular method first, then OCR fallback"""
    print(f"\n Analyzing: {pdf_path}")
    print("=" * 50)

    # First attempt: Regular text extraction
    print(" Attempting regular text extraction...")
    text = extract_text_from_pdf_pdfplumber(pdf_path)

    # Check if we got meaningful text
    if text and len(text.strip()) > 50:  # If we have substantial text
        print(" Text extracted successfully (searchable PDF)")
        return text, "searchable"
    else:
        # Fallback to OCR
        print(" Little/no text found. Switching to OCR for scanned PDF...")
        ocr_text = extract_with_ocr(pdf_path)
        print(" OCR processing completed")
        return ocr_text, "scanned"

def save_results(text, filename, output_dir="extracted_results"):
    """Save extracted text to a file"""
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    output_filename = f"{os.path.splitext(filename)[0]}_extracted.txt"
    output_path = os.path.join(output_dir, output_filename)

    with open(output_path, 'w', encoding='utf-8') as f:
        f.write(text)

    return output_path

# MAIN PROGRAM
def main():
    print(" PDF Text Extractor with OCR Fallback ")
    print("=" * 60)

    # File upload
    uploaded = files.upload()

    if not uploaded:
        print(" No files uploaded. Exiting.")
        return

    results = {}

    for filename in uploaded.keys():
        print(f"\n{'='*60}")
        print(f" PROCESSING: {filename}")
        print(f"{'='*60}")

        # Perform extraction
        extracted_text, pdf_type = smart_pdf_extraction(filename)

        # Save results
        saved_path = save_results(extracted_text, filename)

        # Store results
        results[filename] = {
            'text': extracted_text,
            'type': pdf_type,
            'saved_path': saved_path,
            'char_count': len(extracted_text)
        }

        # Display preview
        print(f"\n EXTRACTION SUMMARY:")
        print(f"   PDF Type: {pdf_type.upper()}")
        print(f"   Characters extracted: {len(extracted_text):,}")
        print(f"   Saved to: {saved_path}")

        print(f"\nTEXT PREVIEW (first 1000 characters):")
        print("-" * 40)
        print(extracted_text[:1000] + "..." if len(extracted_text) > 1000 else extracted_text)
        print("-" * 40)

    # Final summary
    print(f"\n{'='*60}")
    print(" EXTRACTION COMPLETED - SUMMARY")
    print(f"{'='*60}")

    for filename, result in results.items():
        print(f" {filename}:")
        print(f"   Type: {result['type']} | Chars: {result['char_count']:,} | Saved: {result['saved_path']}")

    return results

# Run the program
if __name__ == "__main__":
    extraction_results = main()
```

```python
import re

def clean_text(text):
    # Remove multiple spaces and newlines
    text = re.sub(r'\s+', ' ', text)

    # Remove page numbers (naive: standalone digits)
    text = re.sub(r'\b\d+\b', '', text)

    # Remove weird characters (non-ascii except basic punctuation)
    text = re.sub(r'[^a-zA-Z0-9.,!?;:\'\"()\-\s]', '', text)

    return text.strip()

# Example
raw_text = """
Page 1

This   is     a sample document!!
It has multiple   spaces, 123 numbers, and weird symbols ★★★.

Page 2
More text here...
"""
print("Before Cleaning:\n", raw_text[:300])
print("\nAfter Cleaning:\n", clean_text(raw_text)[:300])
```

```python
import re
import nltk
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.lex_rank import LexRankSummarizer

class AudiobookPreprocessor:
    def __init__(self):
        # Download required NLTK data
        nltk.download('punkt', quiet=True)
        nltk.download('stopwords', quiet=True)
        self.stop_words = set(nltk.corpus.stopwords.words('english'))

    def clean_text(self, text):
        """Clean text for better processing"""
        # Remove extra whitespace
        text = re.sub(r'\s+', ' ', text)
        # Remove standalone page numbers (better version)
        text = re.sub(r'(?i)(^|\n)\s*page\s*\d+\s*($|\n)', '', text)
        # Keep basic punctuation for natural speech
        text = re.sub(r'[^\w\s.,!?;:\'\"()\-]', '', text)
        return text.strip()

    def split_into_sentences(self, text):
        """Split text into natural sentences"""
        sentences = re.split(r'(?<=[.!?]) +', text)
        return [s.strip() for s in sentences if s.strip() and len(s.split()) > 3]

    def remove_stopwords_smart(self, text):
        """Light cleaning keeping sentence structure intact"""
        tokens = nltk.word_tokenize(text)
        # Keep important words but maintain readability
        filtered = [w for w in tokens if w.lower() not in self.stop_words or w in ['.', ',', '!', '?']]
        return ' '.join(filtered)

    def create_audiobook_chunks(self, text, chunk_size=300, overlap_sentences=2):
        """
        Create perfect audiobook chunks with smooth transitions

        Parameters:
        - chunk_size: Target words per chunk (optimal for TTS)
        - overlap_sentences: Sentences to overlap for smooth transitions
        """
        # Step 1: Clean the text
        cleaned_text = self.clean_text(text)

        # Step 2: Split into sentences
        sentences = self.split_into_sentences(cleaned_text)

        # Step 3: Create chunks with overlapping sentences
        chunks = []
        current_chunk = []
        current_word_count = 0

        for i, sentence in enumerate(sentences):
            sentence_word_count = len(sentence.split())

            # If adding this sentence exceeds chunk size (and we have content)
            if current_word_count + sentence_word_count > chunk_size and current_chunk:
                # Save current chunk
                chunks.append(' '.join(current_chunk))

                # Start new chunk with overlap
                overlap_start = max(0, len(current_chunk) - overlap_sentences)
                current_chunk = current_chunk[overlap_start:]
                current_word_count = sum(len(s.split()) for s in current_chunk)

            # Add sentence to current chunk
            current_chunk.append(sentence)
            current_word_count += sentence_word_count

        # Add the final chunk
        if current_chunk:
            chunks.append(' '.join(current_chunk))

        return chunks

    def summarize_long_sections(self, text, max_sentences_per_section=10):
        """Summarize very long sections for better listening"""
        sentences = self.split_into_sentences(text)

        if len(sentences) <= max_sentences_per_section:
            return text

        # Use extractive summarization for long sections
        parser = PlaintextParser.from_string(text, Tokenizer("english"))
        summarizer = LexRankSummarizer()
        summary_sentences = summarizer(parser.document, max_sentences_per_section)

        return ' '.join(str(sentence) for sentence in summary_sentences)

# MAIN USAGE EXAMPLE
def prepare_audiobook_text(full_text, target_chunk_size=300):
    """Complete pipeline for audiobook preparation"""
    processor = AudiobookPreprocessor()

    print(" Processing text for audiobook...")
    print(f"Original text length: {len(full_text):,} characters")

    # Step 1: Clean the entire text
    cleaned_text = processor.clean_text(full_text)
    print(f"After cleaning: {len(cleaned_text):,} characters")

    # Step 2: Create optimal chunks
    chunks = processor.create_audiobook_chunks(cleaned_text, chunk_size=target_chunk_size)

    # Step 3: Process each chunk
    final_chunks = []
    for i, chunk in enumerate(chunks):
        # Summarize if chunk is still too long
        if len(chunk.split()) > target_chunk_size * 1.5:
            chunk = processor.summarize_long_sections(chunk)

        final_chunks.append({
            'chunk_number': i + 1,
            'text': chunk,
            'word_count': len(chunk.split()),
            'duration_estimate': f"{len(chunk.split()) / 150:.1f} mins"  # ~150 words/min
        })

    return final_chunks

# EXAMPLE USAGE
if __name__ == "__main__":
    # Sample text (would be your extracted document text)
    sample_document = """
    Artificial Intelligence is transforming how we interact with technology.
    Machine learning algorithms can now recognize patterns in data that were
    previously invisible to human analysts. This has applications in healthcare,
    finance, and education. However, ethical considerations must be addressed
    as AI becomes more pervasive in our daily lives. The future of AI depends
    on responsible development and deployment. Researchers are working on
    explainable AI systems that can justify their decisions to human users.
    """

    # Process for audiobook
    audiobook_chunks = prepare_audiobook_text(sample_document)

    # Display results
    print(f"\n Generated {len(audiobook_chunks)} audiobook chunks:")
    print("=" * 50)

    for chunk in audiobook_chunks:
        print(f"\nChunk {chunk['chunk_number']}:")
        print(f"Words: {chunk['word_count']} | Est. Duration: {chunk['duration_estimate']}")
        print(f"Preview: {chunk['text'][:100]}...")
```
