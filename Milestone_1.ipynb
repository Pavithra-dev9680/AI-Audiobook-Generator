```python
from google.colab import files

# Upload a TXT file
uploaded = files.upload()

# Read uploaded TXT file
for filename in uploaded.keys():
    print(f"\n--- Contents of {filename} ---\n")
    with open(filename, "r", encoding="utf-8") as f:
        text = f.read()
        print(text[:500])  # show first 500 chars

```

```python
from google.colab import files
import docx

# Upload DOCX file
uploaded = files.upload()

# Read DOCX content
for filename in uploaded.keys():
    doc = docx.Document(filename)
    full_text = []
    for para in doc.paragraphs:
        full_text.append(para.text)
    extracted_text = "\n".join(full_text)
    print(f"\n--- Extracted from {filename} ---\n")
    print(extracted_text[:500])  # preview
```

```python
from google.colab import files
import PyPDF2

# Upload PDF
uploaded = files.upload()

# Extract text with PyPDF2
for filename in uploaded.keys():
    with open(filename, "rb") as f:
        reader = PyPDF2.PdfReader(f)
        text = ""
        for page in reader.pages:
            text += page.extract_text() or ""
        print(f"\n--- Extracted with PyPDF2 from {filename} ---\n")
        print(text[:500])
```

```python
import pdfplumber

for filename in uploaded.keys():
    text = ""
    with pdfplumber.open(filename) as pdf:
        for page in pdf.pages:
            text += page.extract_text() or ""
    print(f"\n--- Extracted with pdfplumber from {filename} ---\n")
    print(text[:500])
```
